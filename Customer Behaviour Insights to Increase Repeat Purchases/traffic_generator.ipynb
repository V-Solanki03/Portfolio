```python
import csv
import time
import random
import os
from datetime import datetime
```

```python
locations = [
    {"location_id": 1, "road_name": "MG Road", "lat": 12.9716, "lon": 77.5946},
    {"location_id": 2, "road_name": "Outer Ring Road", "lat": 12.9121, "lon": 77.6446},
    {"location_id": 3, "road_name": "Electronic City Flyover", "lat": 12.8452, "lon": 77.6600}
]
```

```python
def generate_speed():
    if random.random() < 0.3:  # 30% congestion
        return random.randint(5, 20)
    return random.randint(40, 80)
```

```python
def get_density(speed):
    if speed < 20:
        return "High"
    elif speed < 40:
        return "Medium"
    else:
        return "Low"
```

```python
csv_file = "traffic_input/traffic_data.csv"
file_exists = os.path.isfile(csv_file)
```

```python
with open(csv_file, mode="a", newline="") as file:
    writer = csv.writer(file)

    if not file_exists:
        writer.writerow([
            "timestamp",
            "location_id",
            "latitude",
            "longitude",
            "average_speed_kmph",
            "traffic_density",
            "road_name"
        ])

    loc = locations[0]
    speed = generate_speed()
    density = get_density(speed)

    writer.writerow([
        datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        loc["location_id"],
        loc["lat"],
        loc["lon"],
        speed,
        density,
        loc["road_name"]
    ])
```

```python
with open(csv_file, mode="a", newline="") as file:
    writer = csv.writer(file)

    while True:
        for loc in locations:
            speed = generate_speed()
            density = get_density(speed)

            writer.writerow([
                datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
                loc["location_id"],
                loc["lat"],
                loc["lon"],
                speed,
                density,
                loc["road_name"]
            ])

        file.flush()
        print("Data written at:", datetime.now())
        time.sleep(5)
```
